# Explore the prepared curlew tracks

Run `01_download_data.py` and `02_prepare_features.py` before this notebook. The notebook is only for a short visual check of the prepared modelling table.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
data = pd.read_csv(ROOT / 'data' / 'processed' / 'curlew_ml_table.csv')
print(data.shape)
print(data.groupby('bird_id')['active_movement'].agg(['count', 'mean']))
data.head()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(data['longitude'], data['latitude'], c=data['active_movement'], s=3, alpha=0.5)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Curlew GPS locations and movement label')
plt.show()
```
